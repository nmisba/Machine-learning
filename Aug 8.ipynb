```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("2. Finance_Analytics_Dataset.csv")
```

```python
# show the number of transactions in each category using count plot
sns.countplot(data=df, x="Category")
plt.title("Number of Transactions by Category")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
sns.countplot(data=df, x="PaymentMode")
plt.title("Number of Transactions by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
sns.countplot(data=df, x="TransactionType")
plt.title("Number of Transactions by transaction type")
plt.xlabel("TransactionType")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
sns.barplot(data=df, x="TransactionType", y="Amount")
plt.title("Debit vs Credit Amount")
plt.xlabel("TransactionType")
plt.ylabel("Amount")
plt.show()
```

```python
sns.barplot(data=df, x= "CustomerSegment", y= "Amount")
plt.title("Segment pattern by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Amount")
plt.show()
```

```python
sns.boxplot(data=df,x="Category", y="Amount")
plt.title("Transaction Amount by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.show()
```

```python
plt.pie(df["PaymentMode"].value_counts(),
        labels=df["PaymentMode"].value_counts().index)
plt.title("Payment Mode Usage Distribution")
plt.show()
```
