```python
num_employees = len(df)
print("Number of employees:", num_employees)
```

```python
num_departments = df["Department"].nunique()

print("Number of departments:", num_departments)
print("Departments:", df["Department"].unique())
```

```python
departments = df["Department"].unique()

print("Departments in the dataset:")
for dept in departments:
    print(dept)
```

```python
average_age = df["Age"].mean()

print("Average employee age:", round(average_age, 2))
```

```python
min_age = df["Age"].min()
max_age = df["Age"].max()

print("Minimum employee age:", min_age)
print("Maximum employee age:", max_age)
```

```python
gender_distribution = df["Gender"].value_counts()

print("Gender Distribution:")
print(gender_distribution)
```

```python
department_count = df["Department"].value_counts()

print("Employees in each department:")
print(department_count)
```

```python
count = (df["YearsAtCompany"] > 10).sum()

print("Employees with more than 10 years of experience:", count)
```

```python
count = (df["YearsAtCompany"] > 5).sum()

print("Employees with more than 5 years at the company:", count)
```

```python
average_income = df.groupby("Department")["MonthlyIncome"].mean()

print("Average Monthly Income in Each Department:")
print(average_income.round(2))
```

```python
employees_left = df[df["Attrition"] == "Yes"].groupby("Department").size()

print("Employees who left the company in each department:")
print(employees_left)
```
